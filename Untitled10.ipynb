```python
INF = float('inf')

graph = [
    [0, 4, 2, 0, 0, 0],
    [4, 0, 5,10, 0, 0],
    [2, 5, 0, 0, 3, 0],
    [0,10, 0, 0, 7, 8],
    [0, 0, 3, 7, 0, 6],
    [0, 0, 0, 8, 6, 0]
]

vertices = ['A', 'B', 'C', 'D', 'E', 'F']
V = len(graph)

selected = [False] * V
selected[0] = True

print("Edges in MST:")

cost = 0

for _ in range(V - 1):

    minimum = INF
    x = y = 0

    for i in range(V):
        if selected[i]:
            for j in range(V):
                if (not selected[j] and
                        graph[i][j] != 0 and
                        graph[i][j] < minimum):

                    minimum = graph[i][j]
                    x = i
                    y = j

    print(f"{vertices[x]} - {vertices[y]} = {graph[x][y]}")
    cost += graph[x][y]
    selected[y] = True

print("Total Cost =", cost)
```
