```python

import random

def interpolation_search(arr, key):
    low = 0
    high = len(arr) - 1
    comparisons = 0

    while low <= high and key >= arr[low] and key <= arr[high]:

        comparisons += 1

        if arr[high] == arr[low]:
            if arr[low] == key:
                return low, comparisons
            return -1, comparisons

        pos = low + int(((key - arr[low]) * (high - low)) /
                        (arr[high] - arr[low]))

        if arr[pos] == key:
            return pos, comparisons
        elif arr[pos] < key:
            low = pos + 1
        else:
            high = pos - 1

    return -1, comparisons


sizes = [10000, 50000, 100000]

for size in sizes:

    # Uniformly distributed floating-point numbers
    arr = sorted(random.uniform(0.0, 1000.0) for _ in range(size))

    # Search for an existing value
    key = arr[size // 2]

    index, comparisons = interpolation_search(arr, key)

    print("\nDataset Size:", size)
    print("Found at Index:", index)
    print("Comparisons:", comparisons)
```
