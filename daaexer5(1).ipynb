```python
import random

comparison_count = 0

def min_max_second_dc(arr, low, high):
    global comparison_count

    # Base case: one element
    if low == high:
        return arr[low], float('inf'), arr[low], float('-inf')

    # Base case: two elements
    if high == low + 1:
        comparison_count += 1
        if arr[low] < arr[high]:
            return arr[low], arr[high], arr[high], arr[low]
        else:
            return arr[high], arr[low], arr[low], arr[high]

    # Divide
    mid = (low + high) // 2
    lmin1, lmin2, lmax1, lmax2 = min_max_second_dc(arr, low, mid)
    rmin1, rmin2, rmax1, rmax2 = min_max_second_dc(arr, mid + 1, high)

    # Combine minimums
    comparison_count += 1
    if lmin1 < rmin1:
        overall_min = lmin1
        comparison_count += 1
        second_min = min(lmin2, rmin1)
    else:
        overall_min = rmin1
        comparison_count += 1
        second_min = min(rmin2, lmin1)

    # Combine maximums
    comparison_count += 1
    if lmax1 > rmax1:
        overall_max = lmax1
        comparison_count += 1
        second_max = max(lmax2, rmax1)
    else:
        overall_max = rmax1
        comparison_count += 1
        second_max = max(rmax2, lmax1)

    return overall_min, second_min, overall_max, second_max


# -----------------------------
# Demonstration
# -----------------------------
arr = [3, 1, 7, 4, 9, 2, 8, 5, 6, 0]

comparison_count = 0
mn, sec_mn, mx, sec_mx = min_max_second_dc(arr, 0, len(arr)-1)

print("Array:", arr)
print("Minimum          :", mn)
print("Second Minimum   :", sec_mn)
print("Maximum          :", mx)
print("Second Maximum   :", sec_mx)
print("Comparisons      :", comparison_count)

# -----------------------------
# Performance Analysis
# -----------------------------
print(f"\n{'Size':>8} {'Comparisons':>15}")

for size in [10, 100, 1000, 10000]:
    arr = [random.randint(1, 100000) for _ in range(size)]
    comparison_count = 0
    min_max_second_dc(arr, 0, len(arr)-1)

    print(f"{size:>8} {comparison_count:>15}")
```
