```python
import random

# Partition function
def partition(arr, low, high):
    pivot = arr[high]
    i = low

    for j in range(low, high):
        if arr[j] <= pivot:
            arr[i], arr[j] = arr[j], arr[i]
            i += 1

    arr[i], arr[high] = arr[high], arr[i]
    return i


# QuickSelect (Divide and Conquer)
def quickselect(arr, low, high, k):
    if low <= high:
        pivot_index = partition(arr, low, high)

        if pivot_index == k:
            return arr[pivot_index]
        elif pivot_index > k:
            return quickselect(arr, low, pivot_index - 1, k)
        else:
            return quickselect(arr, pivot_index + 1, high, k)

    return None


# -----------------------------
# Generate random array
# -----------------------------
n = 1000
arr = [random.randint(1, 10000) for _ in range(n)]

# Find Minimum (k = 1)
minimum = quickselect(arr.copy(), 0, n - 1, 0)

# Find Maximum (k = n)
maximum = quickselect(arr.copy(), 0, n - 1, n - 1)

# Find Median (k = n//2)
median = quickselect(arr.copy(), 0, n - 1, n // 2)

# Display Results
print("Array Size :", n)
print("Minimum    :", minimum)
print("Maximum    :", maximum)
print("Median     :", median)

# Verification using sorting
sorted_arr = sorted(arr)

print("\nVerification")
print("Min Correct?    :", minimum == sorted_arr[0])
print("Max Correct?    :", maximum == sorted_arr[-1])
print("Median Correct? :", median == sorted_arr[n // 2])
```
