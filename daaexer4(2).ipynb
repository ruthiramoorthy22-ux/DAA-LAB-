```python
import heapq

def dijkstra(graph, source):
    n = len(graph)

    INF = float('inf')
    dist = [INF] * n
    hops = [INF] * n

    dist[source] = 0
    hops[source] = 0

    # Priority Queue: (distance, hop_count, vertex)
    pq = [(0, 0, source)]

    while pq:
        current_dist, current_hops, u = heapq.heappop(pq)

        if current_dist > dist[u]:
            continue

        for v, weight in graph[u]:
            new_dist = dist[u] + weight
            new_hops = hops[u] + 1

            # Update if shorter distance
            # OR same distance but fewer hops
            if (new_dist < dist[v]) or \
               (new_dist == dist[v] and new_hops < hops[v]):

                dist[v] = new_dist
                hops[v] = new_hops

                heapq.heappush(pq, (new_dist, new_hops, v))

    return dist, hops


# Graph with 6 vertices
graph = {
    0: [(1, 4), (2, 2)],
    1: [(0, 4), (2, 1), (3, 5)],
    2: [(0, 2), (1, 1), (3, 8), (4, 10)],
    3: [(1, 5), (2, 8), (4, 2), (5, 6)],
    4: [(2, 10), (3, 2), (5, 2)],
    5: [(3, 6), (4, 2)]
}

source = 0

distances, hop_counts = dijkstra(graph, source)

print("Vertex\tDistance\tHop Count")
for i in range(len(graph)):
    print(f"{i}\t{distances[i]}\t\t{hop_counts[i]}")
```
