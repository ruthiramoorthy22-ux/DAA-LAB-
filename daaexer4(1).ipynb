```python
import heapq

# Number of nodes
V = 5

# Graph represented as an adjacency matrix
# 0 = Airport
# 1 = Junction A
# 2 = Junction B
# 3 = Bus Terminal
# 4 = Central Station

graph = [
    [0, 10, 3, 0, 0],   # Airport
    [0, 0, 1, 2, 0],    # Junction A
    [0, 4, 0, 8, 2],    # Junction B
    [0, 0, 0, 0, 7],    # Bus Terminal
    [0, 0, 0, 0, 0]     # Central Station
]

def dijkstra(graph, source):
    n = len(graph)
    dist = [float('inf')] * n
    parent = [-1] * n

    dist[source] = 0

    # Priority Queue (distance, node)
    pq = [(0, source)]

    while pq:
        current_dist, u = heapq.heappop(pq)

        if current_dist > dist[u]:
            continue

        for v in range(n):
            if graph[u][v] > 0:
                new_dist = dist[u] + graph[u][v]

                if new_dist < dist[v]:
                    dist[v] = new_dist
                    parent[v] = u
                    heapq.heappush(pq, (new_dist, v))

    return dist, parent


# Function to reconstruct shortest path
def get_path(parent, destination):
    path = []

    while destination != -1:
        path.append(destination)
        destination = parent[destination]

    return path[::-1]


# Run Dijkstra from Airport (Node 0)
dist, parent = dijkstra(graph, 0)

destination = 4
path = get_path(parent, destination)

print("Shortest distances from Airport:")
for i in range(V):
    print(f"Node {i}: {dist[i]} minutes")

print("\nFastest Route from Airport (0) to Central Station (4):")
print(" -> ".join(map(str, path)))

print(f"\nTotal Travel Time: {dist[destination]} minutes")
```
